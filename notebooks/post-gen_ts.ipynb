## Timeseries Generation

This tutorial illustrates how to use the `History` class to convert CESM (versions 1, 2, and 3) output history files to timeseries files.

```python
%load_ext autoreload
%autoreload 2

import os
os.chdir('/glade/u/home/fengzhu/Github/x4c/docsrc/notebooks')
import numpy as np
import x4c
print(x4c.__version__)
```

```python
dirpath = '/glade/derecho/scratch/cmip7/archive/b.e30_beta06.B1850C_LTso.ne30_t232_wgx3.192.wrkflw.1_32'
case = x4c.History(dirpath, casename='b.e30_beta06.B1850C_LTso.ne30_t232_wgx3.192.wrkflw.1')
```

```python
case.paths['atm']['cam.h0a']
```

```python
case.comps_info
```

```python
%%time

case.gen_ts(
    # comps={'atm': ['TS', 'T'], 'ocn': ['SALT']},
    comps={'atm': ['TS']},
    output_dirpath='/glade/campaign/univ/uazn0034/fengzhu/CESM_output/timeseries',
    scratch_dirpath=f'/glade/derecho/scratch/fengzhu/x4c/gen_ts/test',
    timespan=(5501, 5550),
    timestep=50,
    nproc=1,
    overwrite=True,
    compression=1,
)
```

```python
%%time

case.gen_ts(
    # comps={'atm': ['TS', 'T'], 'ocn': ['SALT']},
    comps={'atm': ['TS'], 'ocn': ['SSH']},
    output_dirpath=os.path.join('/glade/campaign/univ/uazn0034/fengzhu/CESM_output/timeseries', case.casename),
    scratch_dirpath=f'/glade/derecho/scratch/fengzhu/x4c/gen_ts/test',
    timespan=(5501, 5501),
    timestep=1,
    nproc=1,
    overwrite=True,
    compression=1,
)
```
