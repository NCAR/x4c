```python
dirpath = os.path.join('/glade/campaign/cesm/development/cross-wg/diagnostic_framework/CESM_output_for_testing/b.e30_beta02.BLT1850.ne30_t232.104')
case = x4c.History(
    dirpath,
    casename='b.e30_beta02.BLT1850.ne30_t232.104',
    cesm_ver=3,
    comps=['ocn'],
    comps_info={'ocn': ('mom6', ['h.native'])}
)

case.gen_ts(
    comps=['ocn'],
    output_dirpath=os.path.join('/glade/derecho/scratch/fengzhu/x4c/gen_ts/test', case.casename),
    staging_dirpath=os.path.join('/glade/derecho/scratch/fengzhu/x4c/gen_ts/test', case.casename),
    timespan=(1, 10),
    timestep=10,
    nproc=4,
    overwrite=True,
)
```

```python
ds = x4c.open_dataset('/glade/derecho/scratch/fengzhu/x4c/gen_ts/test/b.e30_beta02.BLT1850.ne30_t232.104/.bigbang_ocn.h.z.0001-0010/b.e30_beta02.BLT1850.ne30_t232.104.mom6.h.z.agessc.0001-01.nc')
ds
```

```python
vns_in = case.vns['ocn']['h.z']
comp = 'ocn'
hs = 'h.z'
tres = 'month_1'
timespan_tmp = (1, 10)
dir_structure = 'comp/proc/tseries/tres'
staging_dirpath=os.path.join('/glade/derecho/scratch/fengzhu/x4c/gen_ts/test', case.casename)
bigbang_dir = os.path.join(staging_dirpath, f'.bigbang_{comp}.{hs}.{timespan_tmp[0]:04d}-{timespan_tmp[1]:04d}')
bigcrunch_dir = os.path.join(staging_dirpath, dir_structure.replace('comp', comp).replace('tres', tres))
case.bigcrunch(
    comp='ocn',
    hstr='h.z',
    input_dirpath=bigbang_dir,
    output_dirpath=bigcrunch_dir,
    timespan=timespan_tmp,
    overwrite=True,
    nproc=1,
    compression=1,
    vns=vns_in,
)
```
